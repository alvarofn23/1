```python
# Parte 1: Clase Arca
class Arca:
    # Atributos estáticos
    animales = []
    alimentos = []
    agua = 0
    capacidad_maxima = 0

    @classmethod
    def __init__(cls, capacidad_maxima):
        cls.capacidad_maxima = capacidad_maxima
        cls.animales = []
        cls.alimentos = []
        cls.agua = 0

    @classmethod
    def agregar_animal(cls, animal):
        if len(cls.animales) + len(cls.alimentos) < cls.capacidad_maxima:
            cls.animales.append(animal)
        else:
            print("Capacidad máxima alcanzada. No se puede agregar más animales.")

    @classmethod
    def agregar_alimento(cls, alimento):
        if len(cls.animales) + len(cls.alimentos) < cls.capacidad_maxima:
            cls.alimentos.append(alimento)
        else:
            print("Capacidad máxima alcanzada. No se puede agregar más alimentos.")

    @classmethod
    def agregar_agua(cls, cantidad):
        cls.agua += cantidad

    @classmethod
    def alimentar_animal(cls, animal):
        for alimento in cls.alimentos:
            if alimento.es_alimento_adecuado(animal.tipo):
                animal.alimentar()
                alimento.usar(1)
                break
        else:
            print(f"No hay alimento adecuado para {animal.nombre}")

    @classmethod
    def dar_agua(cls, animal):
        if cls.agua > 0:
            animal.dar_agua()
            cls.agua -= 1
        else:
            print("No hay suficiente agua.")

    @staticmethod
    def estado_arca():
        estado = {
            "Numero de animales": len(Arca.animales),
            "Numero de alimentos": len(Arca.alimentos),
            "Cantidad de agua": Arca.agua
        }
        return estado

# Parte 2: Clases Padre (Animal y Alimento)
class Animal:
    def __init__(self, nombre, tipo):
        self.nombre = nombre
        self.tipo = tipo
        self.hambre = 10  # Nivel de hambre inicial
        self.sed = 10     # Nivel de sed inicial

    def alimentar(self):
        if self.hambre > 0:
            self.hambre -= 1
        else:
            print(f"{self.nombre} ya no tiene hambre.")

    def dar_agua(self):
        if self.sed > 0:
            self.sed -= 1
        else:
            print(f"{self.nombre} ya no tiene sed.")

    def estado(self):
        return {"Hambre": self.hambre, "Sed": self.sed}

class Alimento:
    def __init__(self, tipo, cantidad):
        self.tipo = tipo
        self.cantidad = cantidad

    def usar(self, cantidad):
        if self.cantidad >= cantidad:
            self.cantidad -= cantidad
        else:
            print("No hay suficiente alimento.")

    @staticmethod
    def es_alimento_adecuado(tipo_animal):
        return True

# Parte 3: Clases Derivadas
class Perro(Animal):
    def __init__(self, nombre):
        super().__init__(nombre, "perro")

class Gato(Animal):
    def __init__(self, nombre):
        super().__init__(nombre, "gato")

class Heno(Alimento):
    def __init__(self, cantidad):
        super().__init__("heno", cantidad)

class Croquetas(Alimento):
    def __init__(self, cantidad):
        super().__init__("croquetas", cantidad)

# Parte 4: Usando las Clases
# Crear una instancia de Arca
mi_arca = Arca(10)

# Crear instancias de animales
perro1 = Perro("Rex")
gato1 = Gato("Whiskers")

# Crear instancias de alimentos
heno = Heno(5)
croquetas = Croquetas(10)

# Agregar animales y alimentos al arca
mi_arca.agregar_animal(perro1)
mi_arca.agregar_animal(gato1)
mi_arca.agregar_alimento(heno)
mi_arca.agregar_alimento(croquetas)

# Agregar agua
mi_arca.agregar_agua(10)

# Alimentar a los animales y darles agua
mi_arca.alimentar_animal(perro1)
mi_arca.dar_agua(perro1)

# Verificar el estado del arca
print(mi_arca.estado_arca())
```
